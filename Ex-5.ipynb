```python
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read and display an input image
img = cv2.imread('a_crisp_bright_high_resolution_colorful_landsca.png')

# Display image
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# 1. Averaging Filter
averaging = cv2.blur(img, (5, 5))
```

```python
# 2. Weighted Averaging Filter
kernel = np.array([[1, 2, 1],
                   [2, 4, 2],
                   [1, 2, 1]], dtype=np.float32) / 16
weighted = cv2.filter2D(img, -1, kernel)
```

```python
# 3. Gaussian Filter
gaussian = cv2.GaussianBlur(img, (5, 5), 0)
```

```python
# 4. Median Filter
median = cv2.medianBlur(img, 5)
```

```python
# 5. Laplacian Sharpening using Kernel
lap_kernel = np.array([[0, -1, 0],
                       [-1, 5, -1],
                       [0, -1, 0]])

lap_kernel_img = cv2.filter2D(img, -1, lap_kernel)
```

```python
# 6. Laplacian Operator
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)
```

```python
# Convert results to RGB for display
images = [
    img_rgb,
    cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(median, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(lap_kernel_img, cv2.COLOR_BGR2RGB),
    laplacian
]

titles = [
    'Original Image',
    'Averaging Filter',
    'Weighted Averaging',
    'Gaussian Filter',
    'Median Filter',
    'Laplacian Kernel',
    'Laplacian Operator'
]
```

```python
# Display all images
plt.figure(figsize=(14, 10))

for i in range(7):
    plt.subplot(3, 3, i + 1)
    plt.imshow(images[i], cmap='gray' if i == 6 else None)
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
